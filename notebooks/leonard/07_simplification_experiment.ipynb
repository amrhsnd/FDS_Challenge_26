# Does simplification improve accuracy?

Fixed 2×2 comparison: full 289 vs compact 45 features, original vs simpler HGB. Same five grouped folds on 7,996 development battles as notebook 06. Previously opened holdout and its duplicates remain excluded.

Original HGB: 15 leaves, minimum leaf 40, L2=5. Simpler: 7 leaves, minimum leaf 80, L2=10. Both use 200 iterations, learning rate .06, seed 42.

Compact retains roster health/alive summaries, faints, final/late active HP and current/recent boost threat. It removes detailed effects/status/static stats and all-time bench boosts. Recent boosts and team HP remain estimates.

Default reviews saved evidence. Enable the flag below to repeat the 20 fits.


```python
from pathlib import Path
import sys
import json
import pandas as pd
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src/pokemon_challenge").is_dir())
sys.path.insert(0, str(ROOT / "notebooks/leonard"))
from run_simplification_experiment import run
RUN_NEW_EXPERIMENT = False
RUN_FOLDER = ROOT / "outputs/leonard/20261102_122900_385150_simplification"
if RUN_NEW_EXPERIMENT:
    RUN_FOLDER = run()
print("Results:", RUN_FOLDER)
```

```python
results = pd.read_csv(RUN_FOLDER / "cv_results.csv")
paired = pd.read_csv(RUN_FOLDER / "paired_comparison.csv")
print(results[["candidate", "features", "cv_accuracy", "fold_std"]].to_string(index=False))
print(paired[["candidate", "baseline", "delta_pp", "folds_better", "folds_worse"]].to_string(index=False))
```

## Result

Reference: **67.234%**; full/simpler: **66.834%**; compact/original: **65.708%**; compact/simpler: **65.933%**. Both compact candidates lose in all five folds vs reference. Keep full HGB + boost threat. Simplifying helps within the compact set but does not recover the lost accuracy.

These results reject the specific simplifications tested. They do not establish that simplification can never help. Changes in model constraints are bundled; information removal cannot be attributed to a single removed group. Repeated development CV remains exploratory. No holdout/test evaluation or new CSV.
